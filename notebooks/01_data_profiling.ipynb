```python
from datasets import load_dataset

print("Sedang mengunduh dataset dari Hugging Face...")
# Mengunduh dataset wikipedia-id
ds = load_dataset("indonesian-nlp/wikipedia-id", split="train")

# Menyimpan langsung ke format Parquet di folder data/raw/
output_path = "../data/raw/wikipedia_indonesia.parquet"
ds.to_parquet(output_path)

print(f"Dataset berhasil disimpan ke: {output_path}")
```
